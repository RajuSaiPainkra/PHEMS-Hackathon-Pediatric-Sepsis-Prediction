# Pediatric Sepsis Prediction : PHEMS Hackathon
**Development of machine learning algorithms to predict pediatric sepsis in pediatric intensive care unit (PICU)**
<hr><hr>

## Phase 1: Business Understanding (CRISP-DM)

### Project Objective

The primary objective of this challenge is to develop a machine learning algorithm capable of early detection of sepsis in pediatric patients using retrospective physiological data.

The goal is to predict sepsis at least 6 hours before the clinical prediction, leveraging time-series data collected in the Pediatric Intensive Care Unit (PICU). Early detection can significantly improve patient outcomes, reduce mortality rates, and alleviate the economic burden on healthcare systems.

### 2. Business Context
**Overview of Sepsis**
* Sepsis is a life-threatening condition caused by the body’s extreme response to an infection.

* It leads to tissue damage, organ failure, and often death if not treated promptly.

* Pediatric sepsis is a major concern, particularly in PICUs, due to its rapid progression and diagnostic challenges.

* Timely administration of antibiotics is crucial, as delays in treatment increase mortality by 4-8% for every hour of delay.

**Healthcare and Economic Impact**
* Around 30 million global sepsis cases annually result in approximately 6 million deaths.

* Pediatric cases account for an estimated 20 million globally, with high prevalence among children under 5.

* In Europe, pediatric sepsis leads to significant morbidity, mortality, and economic strain on healthcare systems.


**Current Challenges**
* Early detection of sepsis remains difficult due to diverse clinical presentations.

* Lack of timely diagnosis leads to higher rates of organ dysfunction, longer hospital stays, and increased healthcare costs.

* Predictive models for early detection are needed to improve patient outcomes and reduce PICU resource utilization.

### 3. Problem Statement
Develop a machine learning algorithm that:

* Utilizes retrospective physiological and clinical time-series data from pediatric patients in PICUs.

* Predicts the onset of sepsis 6 hours before clinical detection.

* Outputs a positive or negative prediction for each time interval based on the risk of sepsis.


### 4. Key Goals

 **Primary Goal**:
  * Develop a predictive model with high performance in identifying sepsis early.

 **Secondary Goals**:

* Ensure the algorithm is interpretable and actionable for clinical practitioners.

* Optimize execution time and resource efficiency.

* Contribute to improving sepsis outcomes and reducing healthcare costs.

### 5. Success Criteria

**Primary Metric**

* ***Precision-Recall AUC***: A measure of the model’s ability to balance precision and recall, critical for handling imbalanced datasets.

**Secondary Metrics**

* ***Accuracy***: Overall correctness of the model.

* ***F1-Score***: The harmonic mean of precision and recall, ensuring a balance between false positives and false negatives.

**Tiebreaker Criteria**

* Interpretability of the model.

* Execution time and computational resources required to deploy the algorithm.

### 6. Stakeholders

**Primary Stakeholders**

* Pediatric Intensive Care Unit (PICU) clinicians and healthcare providers.

* Hospital administrators focused on improving patient outcomes and resource management.

**Secondary Stakeholders**

* Researchers and data scientists contributing to pediatric sepsis research.

* Healthcare policymakers aiming to reduce pediatric mortality and healthcare costs.

### 7. Constraints and Assumptions
**Constraints**

* Data is retrospective and provided in a time-series format.

* Positive Sepsis Labels have been shifted 6 hours earlier to simulate early detection requirements.

* Limited computational resources for model execution in real-world clinical settings.

**Assumptions**

* The provided dataset is a representative sample of PICU patient populations.

* Early intervention based on the model’s predictions will lead to improved patient outcomes.

* The model’s predictions will integrate seamlessly into existing clinical workflows.

### 8. Risks and Challenges

* **Data Quality Issues**: Missing or inconsistent data in time-series format could affect model accuracy.

* **Class Imbalance**: Sepsis cases may be significantly outnumbered by non-sepsis cases, leading to potential model bias.

* **Clinical Integration**: Ensuring that the model’s predictions are interpretable and actionable for clinicians.

* **Regulatory Compliance**: Adherence to healthcare data privacy and security regulations, such as GDPR.

### 9. Business Benefits

* **Clinical Impact**: Improved patient outcomes through early detection and timely treatment of sepsis.

* **Economic Benefits**: Reduced PICU resource utilization, shorter hospital stays, and lower treatment costs.

* **Research Advancement**: Contribution to pediatric sepsis research through open-source algorithms.
